# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


### ✅ Solution – Task 1
Represent each sentence as a TF-IDF vector, then compute the cosine similarity between every pair of vectors.

````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = ['This is the first document.',
             'This document is the second document.',
             'And this is the third one.',
             'Is this the first document?']

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(sentences)

sim = cosine_similarity(vectors)
pd.DataFrame(sim, index=sentences, columns=sentences).round(3)
````

````python
# Rank the pairs from most to least similar
pairs = [(sentences[i], sentences[j], sim[i, j])
         for i in range(len(sentences)) for j in range(i + 1, len(sentences))]
for a, b, s in sorted(pairs, key=lambda p: -p[2]):
    print(f"{s:.3f}  |  {a}  <->  {b}")
````

**Observations**
- *"This is the first document."* and *"Is this the first document?"* have a similarity of **1.000**: they contain exactly the same words. TF-IDF is a **bag-of-words** method, so it ignores word order and punctuation and cannot tell a statement from a question.
- The pairs involving the second sentence are the next most similar (**0.647**). They share *this, is, the, document*, and *document* even appears twice in sentence 2.
- *"And this is the third one."* is the least similar to the others (**0.308** and **0.225**). It shares only very common words (*this, is, the*), which have low IDF weights.

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


### ✅ Solution – Task 2
Fit TF-IDF on the three sentences and look at the weight of each word in each sentence. The highest weights mark the most *important* (distinctive) words of every sentence.

````python
docs = ['data science is one of the most important fields of science',
        'this is one of the best data science courses',
        'data scientists analyze data']

tfidf2 = TfidfVectorizer()
m = tfidf2.fit_transform(docs)
tfidf_df = pd.DataFrame(m.toarray(), columns=tfidf2.get_feature_names_out(),
                        index=[f"doc{i+1}" for i in range(len(docs))])
tfidf_df.round(3)
````

````python
# IDF of every word: low IDF = appears in many documents = less informative
idf = pd.Series(tfidf2.idf_, index=tfidf2.get_feature_names_out()).sort_values()
print("IDF values:\n", idf.round(3).to_string(), "\n")

# Top-3 most important words per document
for doc_name, row in tfidf_df.iterrows():
    top = row.sort_values(ascending=False).head(3)
    print(doc_name, "->", ", ".join(f"{w} ({v:.3f})" for w, v in top.items()))
````

**Observations**
- **data** appears in all 3 documents, so it has the **lowest IDF (1.0)**: it is not useful for telling the documents apart. Words that appear in only one document (*analyze, best, courses, fields, important, scientists…*) get the highest IDF (1.693).
- Most important words per document:
  - doc1 → **science** and **of** (0.488), then **fields** (0.321). *science* and *of* score high because they appear **twice** in this sentence (high TF).
  - doc2 → **best**, **courses** (0.400). These words are unique to this document.
  - doc3 → **data** (0.641), **analyze**, **scientists** (0.543). *data* still wins here despite its low IDF, because it appears twice in a very short sentence.
- Note that stop words such as *of* and *this* can still get high scores. TF-IDF only reduces the weight of words that are common *in this corpus* of 3 sentences. Removing stop words first would make the "important words" more meaningful.

### ✅ Solution – Task 3
Load the `spoken_words` column, drop empty lines, clean the text with the given `clean_text` function, tokenize it and train a **skip-gram** Word2Vec model (`sg=1`).

````python
import re
import gensim
import nltk
from nltk.tokenize import word_tokenize

simpsons = pd.read_csv("dataset/simpsons_script_lines.csv", low_memory=False)
print("Rows:", len(simpsons))
print("Missing spoken_words:", simpsons["spoken_words"].isna().sum())

simpsons = simpsons.dropna(subset=["spoken_words"])   # lines with no dialogue cannot be cleaned
print("Rows after dropping missing:", len(simpsons))
````

````python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons["clean"] = simpsons["spoken_words"].apply(clean_text)
tokens = [word_tokenize(line) for line in simpsons["clean"]]

print(len(tokens), "tokenized lines")
print("Total tokens:", sum(len(t) for t in tokens))
print("Average tokens per line:", round(sum(len(t) for t in tokens) / len(tokens), 1))
````

````python
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1, seed=42, workers=4)

print("Vocabulary size:", len(Skip_gram_model.wv))
print("Vector for 'homer' (first 10 dims):", Skip_gram_model.wv["homer"][:10].round(3))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



### ✅ Solution – Task 4

````python
for name in ["homer", "marge", "bart"]:
    print(f"Words most similar to '{name}':")
    for word, score in Skip_gram_model.wv.most_similar(name, topn=10):
        print(f"   {word:<15} {score:.3f}")
    print()
````

**Observations**
- The model has clearly learned that these words are **names of characters**: the nearest neighbours of *homer*, *marge* and *bart* are almost all other character names (*abe, grampa, ned, lisa, milhouse, nelson…*).
- Family relations are visible: *marge* is closest to *homer*, and *bart* is closest to *lisa*. *bartholomew* (Bart's full name) appears among the neighbours of *homer*.
- *marge* is also close to terms of endearment (*honey, sweetie, sweetheart*). These are words used in the same contexts, for example when characters address her. Word2Vec captures words that appear in **similar contexts**, not strictly words with the same meaning.
- Similarity scores are all high (0.80–0.87) and close together, typical for a skip-gram model trained with `min_count=1` on a medium-sized corpus (≈1.5 M tokens, 44k-word vocabulary).

Use the wv.doesnt_match() method to :

### ✅ Solution – Task 5

````python
groups = [['jimbo', 'milhouse', 'kearney'],
          ['nelson', 'bart', 'milhouse'],
          ['homer', 'patty', 'selma']]

for g in groups:
    print(f"{g} -> odd one out: {Skip_gram_model.wv.doesnt_match(g)}")
````

````python
# Pairwise similarities explain *why* a word was chosen as the odd one
for g in groups:
    print(g)
    for i in range(3):
        for j in range(i + 1, 3):
            print(f"   sim({g[i]}, {g[j]}) = {Skip_gram_model.wv.similarity(g[i], g[j]):.3f}")
````

**Observations**
- `['jimbo', 'milhouse', 'kearney']` → **milhouse**. *jimbo* and *kearney* are the most similar pair (0.871): they are usually mentioned together as members of the same group of bullies, while *milhouse* is Bart's friend. ✅ sensible.
- `['nelson', 'bart', 'milhouse']` → **nelson**. *bart* and *milhouse* (best friends) are the most similar pair (0.860), so *nelson* is the odd one. The margin is small though (0.822 / 0.824 vs 0.860).
- `['homer', 'patty', 'selma']` → **homer**. *patty* and *selma* are twin sisters who almost always appear together (similarity 0.913, the highest in the whole task). ✅ the clearest result.

`doesnt_match()` computes the mean of the vectors and returns the word furthest from that mean, so the pairwise similarities show exactly why each word was chosen.
