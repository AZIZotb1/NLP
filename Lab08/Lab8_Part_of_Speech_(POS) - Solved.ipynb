# Part-of-speech (POS)

Part-of-speech (POS) tagging is a popular NLP technique which refers to categorizing words in a text (corpus) in correspondence with a particular part of speech, depending on the definition of the word and its context.

## Benefits of POS tagging



- For some applications, we want to focus on certain POS:

E.g.
Nouns are important for information retrival.
Adjectives are important for sentiment analysis.

- POS Taggning is mostly useful in Named Entity Recognition (NER) applications and extracting relations between words.

### pos_tag method

````python
import nltk
nltk.download('averaged_perceptron_tagger_eng', quiet=True)  # needed by pos_tag
nltk.download('punkt', quiet=True); nltk.download('punkt_tab', quiet=True)
````

````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize

text = "I received my new phone broken from Amazon."
tokens = word_tokenize(text)

for word in tokens:
    print(f'word: ({word}), part of speech: ({pos_tag([word])[0][1]})')
````

````python
import spacy
nlp = spacy.load('en_core_web_sm')
````

````python
doc = nlp(u"The quick brown fox jumped over the lazy dog's back.")
````

* To view the coarse POS tag use `token.pos_`
* To view the fine-grained tag use `token.tag_`
* To view the description of either type of tag use `spacy.explain(tag)`

````python
for token in doc:
    print(f'{token.text:{10}} {token.pos_:{8}} {token.tag_:{6}} {spacy.explain(token.tag_)}')
````

In the English language, the same word can have different meanings, even within the same sentence.

For this reason, morphology is important. **spaCy** uses machine learning algorithms to best predict the use of a token in a sentence. Is *"I read books on NLP"* present or past tense?

````python
doc = nlp(u'I read books on NLP.')
r = doc[1]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

````python
doc = nlp(u'I read a book on NLP.')
r = doc[1]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

## Example: Find the adjective words in a sentance


````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize, sent_tokenize
````

````python
text =  "Mobile is very handy. Packed with good performances and nice camera. This is my first iPhone. Very good experience with Apple. Battery is worst. I am hardly getting 5 hours screen on time. Positive thing is I came to know how many hours I was spending on my mobile. Because of poor battery now I am only using mobile for 4 hours a day."

print(text, end='\n\n')

for sent in sent_tokenize(text):
    tokens = word_tokenize(sent)
    part_of_speech = pos_tag(tokens)

    for i in part_of_speech:
        if 'JJ' in i:
            print(i)
    print('============================')
````

### ✅ Solution – Task 1

````python
import spacy
nlp = spacy.load('en_core_web_sm')

sen = nlp(u"I like watching football matches. I do not know how to play it though")

# find the position of "play" instead of counting by hand
idx = [t.text for t in sen].index("play")
print("Index of 'play':", idx, "\n")

# 1. coarse-grained POS tag
print("1. POS tag of 'play'            :", sen[idx].pos_)
# 2. fine-grained POS tag
print("2. Fine-grained tag of 'play'   :", sen[idx].tag_)
# 3. meaning of the fine-grained tag
print("3. Meaning of the fine-grained tag:", spacy.explain(sen[idx].tag_))
````

````python
# 4. text, coarse POS, fine POS and explanation for every token
print(f"{'TEXT':{12}} {'POS':{10}} {'TAG':{8}} EXPLANATION")
print("-" * 60)
for word in sen:
    print(f'{word.text:{12}} {word.pos_:{10}} {word.tag_:{8}} {spacy.explain(word.tag_)}')
````

**Observations**
- *play* is token number **12**: punctuation counts as a token, so the `.` after *matches* takes an index. Finding the index with `.index("play")` avoids counting by hand.
- *play* → coarse tag **VERB**, fine-grained tag **VB** = *verb, base form*. After *how to*, the verb stays in its base (infinitive) form.
- The fine-grained tags carry more detail than the coarse ones. *like* is `VBP` (present tense, not 3rd person), *watching* is `VBG` (gerund / present participle), *matches* is `NNS` (plural noun).
- *do* in *"I do not know"* is tagged **AUX** (auxiliary verb), not VERB, because here it only helps form the negation. *not* is a particle (`PART`).

Compare the explanation of the words “google”  and “table” in the following sentences.

### ✅ Solution – Task 2
The same word form can be a different part of speech depending on context. We tag each sentence and print the analysis of *google* / *table*.

````python
sentences = [
    ("Did you search on Google?", "Google"),
    ("Let us google it.", "google"),
    ("The has placed the food on the table", "table"),
    ("She plans to table a motion for debate in Parliament", "table"),
]

for text, target in sentences:
    sen = nlp(text)
    word = [t for t in sen if t.text == target][0]   # word = sen[#]
    print(text)
    print(f"   {word.text:<8} POS: {word.pos_:<6} TAG: {word.tag_:<5} -> {spacy.explain(word.tag_)}\n")
````

**Observations**

| Sentence | Word | POS | Fine tag | Meaning |
|---|---|---|---|---|
| Did you search on Google? | Google | PROPN | NNP | proper noun (the company) |
| Let us google it. | google | VERB | VB | verb, "to search online" |
| …placed the food on the table | table | NOUN | NN | noun, the furniture |
| She plans to table a motion… | table | VERB | VB | verb, "to propose for discussion" |

The **same word form gets a different part of speech depending on its context**. spaCy decides from the surrounding words, not from the word alone: after a determiner (*the table*) it is a noun, after *to* (*to table*) or *let us* (*google it*) it is a verb. Capitalisation also helps, since *Google* with a capital letter is tagged as a proper noun. This is why POS tagging is done on whole sentences and not on single words. The `pos_tag([word])` loop at the start of the lab tags each word on its own and loses exactly this information.

Note that sentence 3 is ungrammatical as given (*"The has placed…"*, a word is missing), yet *table* is still tagged correctly because the nearby context *on the table* is clear.
